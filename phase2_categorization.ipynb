## Categorize Trends and Update the Google Sheet

```python

import os
import gspread
from dotenv import load_dotenv
from openai import OpenAI

# Load environment variables from .env
load_dotenv()

# Initialize OpenAI client
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

# Connect to Google Sheet
gc = gspread.service_account(filename="credentials.json")
sheet = gc.open("Job_Trends_Agent").sheet1


# Get existing trends from sheet
existing_trends = sheet.col_values(1)  # Assuming trend names are in column A

# Example trending data to categorize
trending_data = [
    "ssc admit card",
    "lic admit card",
    "ibps clerk",
    "rrb notification"
]

def categorize_trend(trend_text):
    """Categorize a trend using GPT-4o-mini"""
    try:
        response = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[{"role": "user", 
           "content": f"Categorize this trend into one of these categories: Education, Government, Exam, Job, Other.\nTrend: {trend_text}"}],
            temperature=0
        )

        category = response.choices[0].message.content.strip()
        return category
    except Exception as e:
        print(f"Error categorizing '{trend_text}': {e}")
        return "Uncategorized"

new_trends_count = 0

for trend in trending_data:
    if trend not in existing_trends:
        category = categorize_trend(trend)
        sheet.append_row([trend, category, "", "", "", ""])
        new_trends_count += 1

print(f"{new_trends_count} new trends added to Google Sheet.")

```
